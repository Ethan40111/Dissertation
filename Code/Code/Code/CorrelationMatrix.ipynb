```python
import pandas as pd
df = pd.read_csv('../Resources/Overall_cleaned.csv')
```

```python
#Just take the columns that can be converted to floats
df_numeric = pd.DataFrame()
for col in df.columns:
    try:    
        df_numeric[col] = pd.to_numeric(df[col])
    except (ValueError, TypeError):
        pass
```

```python
print(df_numeric.columns)
```

```python
corr_matrix = df_numeric.corr()
```

```python
corr_matrix
```

```python
def highlight_high_corr(val):
    color = 'red' if abs(val) > 0.8 else ''
    return f'background-color: {color}'

styled_corr = corr_matrix.round(2).style.applymap(highlight_high_corr)
styled_corr
```

```python
import os

#create Output directory if it doesn't exist (though it should)
os.makedirs('../Output', exist_ok=True)

# Export the styled correlation matrix to HTML
html_str = styled_corr.to_html()
with open('../Output/correlation_matrix.html', 'w', encoding='utf-8') as f:
    f.write(html_str)

print("Correlation matrix exported to ../Output/correlation_matrix.html")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 16))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, annot_kws={'fontsize': 18})
plt.title('Correlation Matrix', fontsize=18)
plt.xticks(rotation=45, ha='right', fontsize=18)
plt.yticks(fontsize=18)
plt.savefig('../Output/correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
#Covariance for these covariates are 1 so investigate why
#Get the number of na values and non na values for each column (investigate whether both columns fill in for each other)
cols = ['Tumor Size Over Time Recode (1988+)',
        'Tumor Size Summary (2016+)']

for c in cols:
    total = len(df[c])
    non_na = df[c].notna().sum()
    na = df[c].isna().sum()
    print(f"{c}: {non_na} non‑NA, {na} NA (out of {total})")
```

```python
col1 = 'Tumor Size Over Time Recode (1988+)'
col2 = 'Tumor Size Summary (2016+)'

both_non_na = df[col1].notna() & df[col2].notna()
both_na = df[col1].isna() & df[col2].isna()
col1_na_col2_non = df[col1].isna() & df[col2].notna()
col1_non_col2_na = df[col1].notna() & df[col2].isna()

print("both non-NA:", both_non_na.sum())
print("both NA:", both_na.sum())
print(f"{col1} NA & {col2} non-NA:", col1_na_col2_non.sum())
print(f"{col1} non-NA & {col2} NA:", col1_non_col2_na.sum())
```
